## 📚 Understanding Overfitting & Underfitting

In supervised learning, one of the key challenges is ensuring that a model generalizes well to new, unseen data. This lab helps you explore two common pitfalls when training models:

---

### 🎯 Overfitting

Overfitting happens when a model learns the training data **too well**, including the **noise or random fluctuations**. This leads to:

- ✅ High training accuracy
- ❌ Low validation accuracy
- Poor generalization to new data

**Example**: A very complex decision tree that perfectly classifies training examples but fails on new passengers.

---

### 🪫 Underfitting

Underfitting occurs when a model is **too simple** to learn the underlying patterns in the data. As a result, it performs poorly on both training and validation sets.

- ❌ Low training accuracy
- ❌ Low validation accuracy
- High bias, low capacity

**Example**: A simple logistic regression that can't capture nonlinear relationships or interactions.

---

### ⚖️ The Sweet Spot (Just Right)

This is the **ideal scenario** where the model captures the essential patterns without memorizing the training data. You’ll observe:

- 👍 Good training accuracy
- 👍 Good validation accuracy
- A model that generalizes well

---

## 👀 What to Observe in the Lab

As you tweak the controls (model type, data size, features, noise level), pay attention to:

| Observation | Likely Scenario |
|-------------|------------------|
| High training accuracy, low validation accuracy | **Overfitting** |
| Low training and validation accuracy | **Underfitting** |
| Both accuracies reasonably high and close | **Sweet Spot** |

Try the following experiments:

1. **Use a simple model + few features + small data** → Underfitting likely.
2. **Use a complex model + small data + high noise** → Overfitting likely.
3. **Use a moderate model + more data + low noise** → Sweet spot possible.

This lab is designed to help you develop **intuition** about how model complexity, data quantity, and noise affect learning, and how to detect and mitigate overfitting/underfitting in practice.

````python
# Method 3: Using Voilà for seamless integration

# Installation (run in terminal):
# pip install voila
# jupyter serverextension enable voila

import ipywidgets as widgets
from IPython.display import display, HTML, Javascript
import json

class OverfittingLabWidget:
    """Interactive widget version of the overfitting lab"""
    
    def __init__(self):
        self.setup_widgets()
        self.create_layout()
    
    def setup_widgets(self):
        """Create interactive widgets"""
        
        # Model selection
        self.model_dropdown = widgets.Dropdown(
            options=[
                ('Simple (Logistic Regression)', 'simple'),
                ('Moderate (Random Forest - 10 trees)', 'moderate'),
                ('Complex (Random Forest - 100 trees)', 'complex'),
                ('Over-complex (Random Forest - 500 trees)', 'overcomplex')
            ],
            value='simple',
            description='Model Type:'
        )
        
        # Data size selection
        self.data_size_dropdown = widgets.Dropdown(
            options=[
                ('Small (50 samples)', 'small'),
                ('Medium (200 samples)', 'medium'),
                ('Large (500 samples)', 'large'),
                ('Full Dataset', 'full')
            ],
            value='small',
            description='Data Size:'
        )
        
        # Feature complexity
        self.features_dropdown = widgets.Dropdown(
            options=[
                ('Basic (pclass, age, fare)', 'basic'),
                ('Extended (+ sibsp, parch)', 'extended'),
                ('Engineered (+ family_size, is_alone)', 'engineered')
            ],
            value='basic',
            description='Features:'
        )
        
        # Noise level
        self.noise_slider = widgets.FloatSlider(
            value=0.0,
            min=0.0,
            max=0.3,
            step=0.05,
            description='Noise Level:',
            readout_format='.0%'
        )
        
        # Run button
        self.run_button = widgets.Button(
            description='🚀 Run Experiment',
            button_style='primary',
            layout=widgets.Layout(width='200px', height='40px')
        )
        
        # Results display
        self.results_output = widgets.Output()
        
        # Set up event handlers
        self.run_button.on_click(self.run_experiment)
    
    def create_layout(self):
        """Create the widget layout"""
        
        # Header
        header = widgets.HTML("""
        <div style="text-align: center; padding: 20px; background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); 
                    color: white; border-radius: 10px; margin-bottom: 20px;">
            <h1>🚢 Lab 1: Overfitting & Underfitting</h1>
            <p>Interactive lab demonstrating how model complexity affects performance</p>
        </div>
        """)
        
        # Concepts explanation
        concepts = widgets.HTML("""
        <div style="display: grid; grid-template-columns: repeat(auto-fit, minmax(250px, 1fr)); gap: 15px; margin-bottom: 20px;">
            <div style="background: linear-gradient(135deg, #a8edea, #fed6e3); padding: 15px; border-radius: 10px;">
                <h3>🎯 Overfitting</h3>
                <p>Model learns training data too well, including noise. High training accuracy, low validation accuracy.</p>
            </div>
            <div style="background: linear-gradient(135deg, #ffecd2, #fcb69f); padding: 15px; border-radius: 10px;">
                <h3>📉 Underfitting</h3>
                <p>Model is too simple to capture patterns. Poor performance on both training and validation.</p>
            </div>
            <div style="background: linear-gradient(135deg, #d299c2, #fef9d7); padding: 15px; border-radius: 10px;">
                <h3>⚖️ Sweet Spot</h3>
                <p>Optimal complexity that generalizes well to new data while capturing essential patterns.</p>
            </div>
        </div>
        """)
        
        # Controls section
        controls_box = widgets.VBox([
            widgets.HTML("<h3>🔧 Experiment Controls</h3>"),
            widgets.HBox([self.model_dropdown, self.data_size_dropdown]),
            widgets.HBox([self.features_dropdown, self.noise_slider]),
            widgets.HBox([self.run_button], layout=widgets.Layout(justify_content='center'))
        ], layout=widgets.Layout(border='2px solid #3498db', padding='15px', border_radius='10px'))
        
        # Main layout
        self.layout = widgets.VBox([
            header,
            concepts,
            controls_box,
            self.results_output
        ])
    
    def simulate_model(self, model_type, data_size, features, noise_level):
        """Simulate model performance"""
        
        # Configuration from JSON
        configs = {
            'simple': {'base_train': 0.75, 'base_val': 0.73},
            'moderate': {'base_train': 0.85, 'base_val': 0.82},
            'complex': {'base_train': 0.92, 'base_val': 0.85},
            'overcomplex': {'base_train': 0.98, 'base_val': 0.79}
        }
        
        size_effects = {'small': 0.85, 'medium': 0.95, 'large': 1.0, 'full': 1.02}
        feature_effects = {'basic': 0.95, 'extended': 1.0, 'engineered': 1.05}
        
        import random
        config = configs[model_type]
        size_adj = size_effects[data_size]
        feature_adj = feature_effects[features]
        
        train_acc = config['base_train'] * size_adj * feature_adj + random.uniform(-0.05, 0.05)
        val_acc = config['base_val'] * size_adj * feature_adj + random.uniform(-0.05, 0.05) - noise_level * 0.3
        
        # Ensure realistic bounds
        train_acc = max(0.5, min(1.0, train_acc))
        val_acc = max(0.5, min(0.95, val_acc))
        
        # Ensure overfitting for complex models with small data
        if model_type == 'overcomplex' or (model_type == 'complex' and data_size == 'small'):
            train_acc = max(train_acc, val_acc + 0.05)
        
        return train_acc, val_acc
    
    def analyze_results(self, train_acc, val_acc):
        """Analyze and interpret results"""
        gap = train_acc - val_acc
        
        if train_acc < 0.7 and val_acc < 0.7:
            status = "Underfitting"
            interpretation = "📉 Both accuracies are low. Try a more complex model or better features."
        elif gap < 0.03:
            status = "Good Fit"
            interpretation = "✅ Excellent! Good balance between training and validation performance."
        elif gap < 0.08:
            status = "Slight Overfit"
            interpretation = "⚠️ Minor overfitting. Consider regularization."
        elif gap < 0.15:
            status = "Overfitting"
            interpretation = "🔴 Clear overfitting. Reduce complexity or add more data."
        else:
            status = "Severe Overfit"
            interpretation = "🚨 Severe overfitting! Model memorized training data."
        
        return status, interpretation, gap
    
    def run_experiment(self, button):
        """Run the experiment when button is clicked"""
        
        with self.results_output:
            self.results_output.clear_output()
            
            # Show loading
            print("🔄 Running experiment...")
            
            # Get parameters
            model_type = self.model_dropdown.value
            data_size = self.data_size_dropdown.value
            features = self.features_dropdown.value
            noise_level = self.noise_slider.value
            
            # Simulate model
            train_acc, val_acc = self.simulate_model(model_type, data_size, features, noise_level)
            status, interpretation, gap = self.analyze_results(train_acc, val_acc)
            
            # Display results
            results_html = f"""
            <div style="margin-top: 20px;">
                <h3>📊 Results</h3>
                <div style="display: grid; grid-template-columns: repeat(auto-fit, minmax(150px, 1fr)); gap: 15px; margin-bottom: 20px;">
                    <div style="background: #3498db; color: white; padding: 15px; border-radius: 10px; text-align: center;">
                        <div style="font-size: 1.8em; font-weight: bold;">{train_acc:.1%}</div>
                        <div>Training Accuracy</div>
                    </div>
                    <div style="background: #e74c3c; color: white; padding: 15px; border-radius: 10px; text-align: center;">
                        <div style="font-size: 1.8em; font-weight: bold;">{val_acc:.1%}</div>
                        <div>Validation Accuracy</div>
                    </div>
                    <div style="background: #f39c12; color: white; padding: 15px; border-radius: 10px; text-align: center;">
                        <div style="font-size: 1.8em; font-weight: bold;">{gap:.1%}</div>
                        <div>Performance Gap</div>
                    </div>
                    <div style="background: #27ae60; color: white; padding: 15px; border-radius: 10px; text-align: center;">
                        <div style="font-size: 1.2em; font-weight: bold;">{status}</div>
                        <div>Model Status</div>
                    </div>
                </div>
                <div style="background: #f8f9fa; border-left: 4px solid #e74c3c; padding: 15px; border-radius: 0 10px 10px 0;">
                    <h4>🔍 Interpretation</h4>
                    <p>{interpretation}</p>
                </div>
            </div>
            """
            
            display(HTML(results_html))
    
    def display(self):
        """Display the widget"""
        display(self.layout)

# Usage in Jupyter:
def create_interactive_lab():
    """Create and display the interactive lab"""
    lab = OverfittingLabWidget()
    lab.display()
    return lab

# To use: uncomment the line below
````
