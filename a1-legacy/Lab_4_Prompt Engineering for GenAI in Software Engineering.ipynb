### ✅ Basic Prompt with Role Separation (System + User)

This is the foundational prompt structure for interacting with modern chat-based language models.

- The **System message** defines the role and behavior of the assistant.
- The **User message** describes the task and provides the input.

This format encourages consistent behavior and better alignment with user expectations.

In this lab, we ask the AI to act as a pair programmer and optimize a simple Python function. We provide clear instructions and the function we want improved.


```python
messages = [
    {
        "role": "system",
        "content": """
You are an expert AI pair programmer. Your job is to help developers improve the efficiency, clarity, and performance of their code.
Always consider best practices in algorithmic optimization, readability, and edge case handling.
Provide clear explanations and suggest alternative implementations when relevant.
"""
    },
    {
        "role": "user",
        "content": """
I need help optimizing a Python function. I don't have specific metrics, but I want to make this function faster and cleaner.

Please:
- Identify performance bottlenecks, if any.
- Suggest improvements with reasoning.
- Provide a revised version of the function.
- Optionally, mention trade-offs or assumptions if applicable.

Here’s the function I want to improve:

def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True
"""
    }
]
messages
```

```python
# Example call
response = client.chat.completions.create(
    model="gpt-4o",
    messages=messages,
    temperature=0.7
)

print(response.choices[0].message.content)
```

### 🌳 Tree-of-Thought Prompt: Step-by-Step Reasoning

This version introduces a *Tree-of-Thought (ToT)* structure, where the model is guided to reason through the task in **discrete, logical steps**:

1. Analyze the current code
2. Generate multiple options
3. Evaluate trade-offs
4. Select and implement the best one

This mirrors how a skilled engineer might approach a problem and leads to **more deliberate and thoughtful outputs**, especially in complex coding scenarios.

ToT encourages exploration and structured comparison—moving beyond a “single-shot” answer toward **divergent thinking** and **decision-based generation**.


```python
messages = [
    {
        "role": "system",
        "content": """
You are an expert AI pair programmer. Use Tree-of-Thought (ToT) reasoning to break down complex optimization tasks into a sequence of logical steps.
Your job is to identify potential improvements to a function by analyzing it carefully, generating multiple options, and evaluating the trade-offs.
"""
    },
    {
        "role": "user",
        "content": """
Please optimize the following function by following these steps:
1. Analyze the current function and identify bottlenecks or inefficiencies.
2. Think of at least 2–3 alternative ways to implement the logic.
3. Evaluate trade-offs in terms of speed, readability, and complexity.
4. Choose the best solution and implement it.

Function:

def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True
"""
    }
]
messages
```

```python
# Example call
response = client.chat.completions.create(
    model="gpt-4",
    messages=messages,
    temperature=0.7
)

print(response.choices[0].message.content)
```

### 🔁 Tree-of-Thought + Multi-Pass Refinement: Layered Thinking

This version adds **Multi-Pass Refinement** to the Tree-of-Thought approach.

The model is asked to:
- Conduct an **initial pass** to analyze and generate ideas.
- Do a **second pass** to compare and refine those ideas.
- Perform a **final pass** to implement the chosen solution cleanly and clearly.

This simulates a **real engineering workflow**, where solutions are improved over multiple iterations. It encourages:
- Deeper evaluation of trade-offs
- Better-structured code
- More explainable and maintainable outputs

This strategy is ideal for tasks that require **thoughtfulness, correctness, and creativity**—like refactoring legacy code or designing algorithms.


```python
messages = [
    {
        "role": "system",
        "content": """
You are an expert AI engineer using Tree-of-Thought reasoning with Multi-Pass refinement.
Your task is to methodically improve the given function by conducting multiple analytical passes, considering correctness, performance, and clarity.
"""
    },
    {
        "role": "user",
        "content": """
I need help improving this function. Please perform the following:

**Pass 1 – Analysis:**
- Identify inefficiencies and evaluate the current structure.
- List at least 3 alternative strategies or optimizations.

**Pass 2 – Evaluation:**
- Compare each strategy for performance and readability.
- Choose the best candidate based on trade-offs.

**Pass 3 – Implementation:**
- Provide the improved version of the function.
- Add inline comments explaining the improvements.

Function to optimize:

def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True
"""
    }
]
messages
```

```python
# Example call
response = client.chat.completions.create(
    model="gpt-4",
    messages=messages,
    temperature=0.7
)

print(response.choices[0].message.content)
```
